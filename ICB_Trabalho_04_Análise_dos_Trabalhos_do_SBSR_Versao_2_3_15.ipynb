```python
### Grafo 01 - var_url_inicial = "https://www.sbsr.com.br/sbsr-2027/page/8162-inicio"

# url usada nesta célula - var_url_grafo_02 = var_url_inicial

from pathlib import Path
import os
import requests
from bs4 import BeautifulSoup
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from rich.console import Console
from rich.table import Table

# Define o caminho fixo para salvar os arquivos
var_caminho_grafo_01 = Path(r"C:\Users\profs\ICD_Trabalho_04_Análise_dos_Trabalhos_do_SBSR\grafo_01")
var_caminho_grafo_01.mkdir(parents=True, exist_ok=True)

def obter_pasta_downloads():
    var_home = os.path.expanduser("~")
    var_downloads = os.path.join(var_home, "Downloads")
    if not os.path.exists(var_downloads):
        return var_home
    return var_downloads

def coletar_urls(var_url_grafo_01):
    var_urls = set()
    try:
        var_resposta = requests.get(var_url_grafo_01)
        var_resposta.raise_for_status()
        var_soup = BeautifulSoup(var_resposta.text, 'html.parser')
        var_urls.add(var_url_grafo_01)
        for var_link in var_soup.find_all('a', href=True):
            var_href = var_link['href']
            if var_href.startswith('http'):
                var_urls.add(var_href)
    except Exception as var_e:
        print(f"Erro ao acessar {var_url_grafo_01}: {var_e}")
    return list(var_urls)

def salvar_urls_em_arquivos(var_urls):
    global var_df_grafo_01_urls
    var_df_grafo_01_urls = pd.DataFrame({
        'id': range(len(var_urls)),
        'url_grafo_01': var_urls
    })
    var_caminho_csv = var_caminho_grafo_01 / "df_grafo_01_urls.csv"
    var_caminho_tsv = var_caminho_grafo_01 / "df_grafo_01_urls.tsv"

    var_df_grafo_01_urls.to_csv(var_caminho_csv, index=False, encoding='utf-8')
    var_df_grafo_01_urls.to_csv(var_caminho_tsv, sep='\t', index=False, encoding='utf-8')

    print(f"URLs salvas em CSV: {var_caminho_csv}")
    print(f"URLs salvas em TSV: {var_caminho_tsv}")

    # Exibir DataFrame colorido com rich
    console = Console()
    table = Table(show_header=True, header_style="bold black on light_sky_blue1")

    table.add_column("id", style="black on light_sky_blue1", max_width=5)
    table.add_column("url_grafo_01", style="black on light_sky_blue1", overflow="fold", max_width=100)

    for i, row in var_df_grafo_01_urls.iterrows():
        style = "on white" if i % 2 == 0 else "on grey93"
        url_link = f"[link={row['url_grafo_01']}]{row['url_grafo_01']}[/link]"
        table.add_row(str(row['id']), url_link, style=style)

    console.print(table)

def construir_grafo(var_urls):
    var_G = nx.DiGraph()
    for var_i, var_url in enumerate(var_urls):
        var_G.add_node(var_i, url=var_url)
    for var_i in range(1, len(var_urls)):
        var_G.add_edge(0, var_i)
    return var_G

def desenhar_grafo(var_G):
    var_pos = nx.spring_layout(var_G, seed=42)
    plt.figure(figsize=(14, 8))

    var_nodes = list(var_G.nodes())
    var_colors = ['green' if node == 0 else 'red' for node in var_nodes]
    nx.draw_networkx_nodes(var_G, var_pos, node_color=var_colors, node_size=800)

    var_labels = {var_node: str(var_node) for var_node in var_nodes}
    nx.draw_networkx_labels(var_G, var_pos, var_labels, font_color='white', font_weight='bold')

    nx.draw_networkx_edges(var_G, var_pos, arrows=True, edge_color='blue', width=2)

    var_urls = nx.get_node_attributes(var_G, 'url')
    var_legenda = "Legenda:\n\n"
    for var_i in sorted(var_urls.keys()):
        var_legenda += f"{var_i}: {var_urls[var_i]}\n"

    plt.gca().text(1, 1, var_legenda, fontsize=10, va='center', ha='left', transform=plt.gca().transAxes,
                   bbox=dict(facecolor='yellow', edgecolor='black', boxstyle='round,pad=0.3'))

    plt.title("Grafo 01 - url principal do SBSR e links externos")
    plt.axis('off')
    plt.tight_layout()

    var_caminho_arquivo = var_caminho_grafo_01 / "grafo_01_url_principal_SBSR_links_externos_colorido.tiff"
    plt.savefig(var_caminho_arquivo, format='tiff', dpi=300)
    print("\nArquivo salvo com sucesso em:", var_caminho_arquivo)
    plt.show()

# Defina a URL inicial do grafo
var_url_inicial = "https://www.sbsr.com.br/sbsr-2027/page/8162-inicio"
var_url_grafo_01 = var_url_inicial

# Executa o processo completo
var_urls = coletar_urls(var_url_grafo_01)
salvar_urls_em_arquivos(var_urls)
var_grafo = construir_grafo(var_urls)
desenhar_grafo(var_grafo)
```

```python
# Grafo 2 

from pathlib import Path
import os
import requests
from bs4 import BeautifulSoup
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from urllib.parse import urlparse, urljoin
from rich.console import Console
from rich.table import Table

# Define o caminho fixo para salvar os arquivos
var_caminho_grafo_02 = Path(r"C:\Users\profs\ICD_Trabalho_04_Análise_dos_Trabalhos_do_SBSR\grafo_02")
var_caminho_grafo_02.mkdir(parents=True, exist_ok=True)

def obter_pasta_downloads():
    var_home = os.path.expanduser("~")
    var_downloads = os.path.join(var_home, "Downloads")
    return var_downloads if os.path.exists(var_downloads) else var_home

def eh_link_interno(var_url_grafo_02, var_link):
    var_base_dominio = urlparse(var_url_grafo_02).netloc
    var_link_parsed = urlparse(var_link)
    return (var_link_parsed.netloc == "" or var_link_parsed.netloc == var_base_dominio)

def coletar_links_internos(var_url_grafo_02):
    var_urls = set()
    try:
        var_resposta = requests.get(var_url_grafo_02)
        var_resposta.raise_for_status()
        var_soup = BeautifulSoup(var_resposta.text, 'html.parser')
        var_urls.add(var_url_grafo_02)
        for var_tag in var_soup.find_all('a', href=True):
            var_href = var_tag['href']
            var_href_completo = urljoin(var_url_grafo_02, var_href)
            if eh_link_interno(var_url_grafo_02, var_href_completo):
                var_urls.add(var_href_completo)
    except Exception as var_e:
        print(f"Erro ao acessar {var_url_grafo_02}: {var_e}")
    return list(var_urls)

def salvar_urls_em_arquivos(var_urls):
    global var_df_grafo_02_urls
    var_df_grafo_02_urls = pd.DataFrame({
        'id': range(len(var_urls)),
        'url_grafo_02': var_urls
    })

    var_caminho_csv = var_caminho_grafo_02 / "df_grafo_02_urls.csv"
    var_caminho_tsv = var_caminho_grafo_02 / "df_grafo_02_urls.tsv"

    var_df_grafo_02_urls.to_csv(var_caminho_csv, index=False, encoding='utf-8')
    var_df_grafo_02_urls.to_csv(var_caminho_tsv, sep='\t', index=False, encoding='utf-8')

    print(f"URLs salvas em CSV: {var_caminho_csv}")
    print(f"URLs salvas em TSV: {var_caminho_tsv}")

    # Exibir DataFrame colorido com rich e links clicáveis
    console = Console()
    table = Table(show_header=True, header_style="bold black on light_sky_blue1")

    table.add_column("id", style="black on light_sky_blue1")
    table.add_column("url_grafo_02", style="black on light_sky_blue1", overflow="fold")

    for i, row in var_df_grafo_02_urls.iterrows():
        style = "on white" if i % 2 == 0 else "on grey93"
        url_link = f"[link={row['url_grafo_02']}]{row['url_grafo_02']}[/link]"
        table.add_row(str(row['id']), url_link, style=style)

    console.print(table)

def construir_grafo(var_urls):
    var_G = nx.DiGraph()
    for var_i, var_url in enumerate(var_urls):
        var_G.add_node(var_i, url=var_url)
    for var_i in range(1, len(var_urls)):
        var_G.add_edge(0, var_i)
    return var_G

def desenhar_grafo(var_G):
    var_pos = nx.spring_layout(var_G, seed=42)
    plt.figure(figsize=(14, 8))

    var_nodes = list(var_G.nodes())
    var_colors = ['green' if node == 0 else 'red' for node in var_nodes]
    nx.draw_networkx_nodes(var_G, var_pos, node_color=var_colors, node_size=800)

    var_labels = {var_node: str(var_node) for var_node in var_nodes}
    nx.draw_networkx_labels(var_G, var_pos, var_labels, font_color='white', font_weight='bold')

    nx.draw_networkx_edges(var_G, var_pos, arrows=True, edge_color='blue', width=2)

    var_urls = nx.get_node_attributes(var_G, 'url')
    var_legenda = "Legenda:\n"
    for var_i in sorted(var_urls.keys()):
        var_legenda += f"{var_i}: {var_urls[var_i]}\n"

    plt.gca().text(1, 1, var_legenda, fontsize=10, va='center', ha='left',
                   transform=plt.gca().transAxes,
                   bbox=dict(facecolor='yellow', edgecolor='black', boxstyle='round,pad=0.3'))

    plt.title("Grafo 02 - url principal do SBSR e Links Internos")
    plt.axis('off')
    plt.tight_layout()

    var_caminho_arquivo = var_caminho_grafo_02 / "grafo_02_url_principal_SBSR_links_internos_colorido.tiff"
    plt.savefig(var_caminho_arquivo, format='tiff', dpi=300)
    print("\nArquivo salvo com sucesso em:", var_caminho_arquivo)
    plt.show()

# Defina a URL inicial do grafo 02
var_url_inicial = "https://www.sbsr.com.br/sbsr-2027/page/8162-inicio"
var_url_grafo_02 = var_url_inicial

# Executar o processo
var_urls = coletar_links_internos(var_url_grafo_02)
salvar_urls_em_arquivos(var_urls)
var_grafo = construir_grafo(var_urls)
desenhar_grafo(var_grafo)

```

```python
###  código Passo 2

import datetime
from pathlib import Path
import requests
import chardet
import pandas as pd
from bs4 import BeautifulSoup
from rich.console import Console
from rich.table import Table

def detectar_e_corrigir_bytes(var_input_bytes):
    var_resultado = chardet.detect(var_input_bytes)
    var_enc = var_resultado['encoding']
    try:
        return var_input_bytes.decode(var_enc)
    except (UnicodeDecodeError, TypeError):
        return var_input_bytes.decode('utf-8', errors='ignore')

def corrigir_acentos(var_texto):
    if isinstance(var_texto, bytes):
        return detectar_e_corrigir_bytes(var_texto)
    elif isinstance(var_texto, str):
        return var_texto
    else:
        return str(var_texto)

var_caminho_Trabalho_04_jupyter = Path(r"C:\Users\profs\ICD_Trabalho_04_Análise_dos_Trabalhos_do_SBSR")
var_caminho_Trabalho_04_jupyter.mkdir(parents=True, exist_ok=True)

var_url_metadados = "http://marte.sid.inpe.br/ibi/dpi.inpe.br/banon/2003/12.10.19.30.54?ibiurl.requiredsite=marte.sid.inpe.br&forcehistorybackflag=0"

var_response = requests.get(var_url_metadados)

if var_response.status_code == 200:
    var_encoding_detectado = chardet.detect(var_response.content)['encoding']
    var_response.encoding = var_encoding_detectado if var_encoding_detectado else 'utf-8'
    var_html = var_response.text
else:
    raise Exception(f"Erro ao baixar página: {var_response.status_code}")

var_soup = BeautifulSoup(var_html, 'html.parser')
var_metatags = var_soup.find_all('meta')

var_dados_meta = []
for var_tag in var_metatags:
    var_nome = var_tag.get('name') or var_tag.get('property')
    var_conteudo = var_tag.get('content')
    if var_nome and var_conteudo:
        var_nome_corrigido = corrigir_acentos(var_nome)
        var_conteudo_corrigido = corrigir_acentos(var_conteudo)
        var_dados_meta.append({'nome': var_nome_corrigido, 'conteudo': var_conteudo_corrigido})

var_df_meta = pd.DataFrame(var_dados_meta)

var_arquivo_csv = var_caminho_Trabalho_04_jupyter / 'metadados_SBSR.csv'

# Tentar salvar o arquivo normalmente
try:
    var_df_meta.to_csv(var_arquivo_csv, index=False, encoding='utf-8-sig')
    print(f"Arquivo salvo em: {var_arquivo_csv}")
except PermissionError:
    print(f"Permissão negada para salvar em {var_arquivo_csv}. Salvando com nome alternativo.")
    var_data_hora = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    var_arquivo_csv_v2 = var_caminho_Trabalho_04_jupyter / f"metadados_SBSR_{var_data_hora}.csv"
    var_df_meta.to_csv(var_arquivo_csv_v2, index=False, encoding='utf-8-sig')
    print(f"Arquivo salvo em: {var_arquivo_csv_v2}")
except Exception as e:
    print(f"Erro ao salvar o arquivo: {e}")

# Configurar pandas para imprimir DataFrame completo
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Ler o arquivo CSV salvo (a partir do nome original ou alternativo)
try:
    var_df_lido = pd.read_csv(var_arquivo_csv, encoding='utf-8-sig')
except FileNotFoundError:
    var_df_lido = pd.read_csv(var_arquivo_csv_v2, encoding='utf-8-sig')

# Imprimir DataFrame usando rich com fundo azul claro e texto preto

console = Console()
table = Table(show_header=True, header_style="bold black on light_sky_blue1")

table.add_column("nome", no_wrap=False)
table.add_column("conteudo", no_wrap=False, max_width=80)

for i, row in var_df_lido.iterrows():
    style = "on white" if i % 2 == 0 else "on light_gray"
    table.add_row(str(row['nome']), str(row['conteudo']), style=style)



# Opcional: listar métodos e atributos da resposta
#print(dir(var_response))
```

```python
## DATAFRAME capa.htm

import requests
from bs4 import BeautifulSoup, Tag
from collections import defaultdict
from urllib.parse import urlparse
import difflib
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import squareform

# Dicionário com URLs dos congressos
var_urls = {
    "1978": "http://marte.sid.inpe.br/col/dpi.inpe.br/marte@80/2008/09.25.20.31/doc/capa.htm",
    "1982": "http://marte.sid.inpe.br/col/dpi.inpe.br/marte@80/2008/10.01.23.18/doc/capa.htm",
    "1984": "http://marte.sid.inpe.br/col/ltid.inpe.br/sbsr/2009/04.03.17.43/doc/capa.htm",
    "1986": "http://marte.sid.inpe.br/col/dpi.inpe.br/marte@80/2009/03.03.11.29/doc/capa.htm",
    "1988": "http://marte.sid.inpe.br/col/dpi.inpe.br/marte@80/2008/07.28.17.53/doc/capa.htm",
    "1990": "http://marte.sid.inpe.br/col/dpi.inpe.br/marte@80/2008/08.26.21.29/doc/capa.htm",
    "1993": "http://marte.sid.inpe.br/col/dpi.inpe.br/marte@80/2008/06.23.20.41/doc/capa.htm",
    "1996": "http://marte.sid.inpe.br/col/dpi.inpe.br/lise/2004/08.25.13.20/doc/capa.htm",
    "1998": "http://marte.sid.inpe.br/col/dpi.inpe.br/lise/2004/08.11.17.13/doc/capa.htm",
    "2001": "http://marte.sid.inpe.br/col/dpi.inpe.br/lise/2001/12.28.15.53/doc/capa.htm",
    "2003": "http://marte.sid.inpe.br/col/ltid.inpe.br/sbsr/2003/03.28.12.35/doc/capa.htm",
    "2005": "http://marte.sid.inpe.br/col/ltid.inpe.br/sbsr/2005/02.12.16.31/doc/capa.htm",
    "2007": "http://marte.sid.inpe.br/col/dpi.inpe.br/sbsr@80/2007/01.31.19.10/doc/capa.htm",
    "2009": "http://marte.sid.inpe.br/col/dpi.inpe.br/sbsr@80/2009/03.02.18.19/doc/capa.htm",
    "2011": "http://marte.sid.inpe.br/col/dpi.inpe.br/marte/2011/09.28.16.42/doc/capa.htm",
    "2013": "http://marte2.sid.inpe.br/rep/dpi.inpe.br/marte2/2013/05.28.22.25/capa.htm",
    "2015": "http://marte2.sid.inpe.br/rep/sid.inpe.br/marte2/2015/05.31.21.54/capa.htm",
    "2017": "http://marte2.sid.inpe.br/rep/sid.inpe.br/marte2/2017/09.25.14.55/capa.htm",
    "2019": "http://marte2.sid.inpe.br/rep/sid.inpe.br/marte2/2019/11.08.12.52/capa.htm",
    "2023": "http://marte2.sid.inpe.br/rep/sid.inpe.br/marte2/2023/05.18.17.15/capa.htm",
    "2025": "http://marte2.sid.inpe.br/rep/sid.inpe.br/marte2/2025/09.17.19.06/capa.htm"
}

def var_extrair_estrutura_tag(var_tag, var_nivel=0):
    if not isinstance(var_tag, Tag):
        return ""
    var_estrutura = var_tag.name
    var_filhos = [var_extrair_estrutura_tag(filho, var_nivel+1) for filho in var_tag.children if isinstance(filho, Tag)]
    if var_filhos:
        var_estrutura += "(" + ",".join(var_filhos) + ")"
    return var_estrutura

def var_obter_estrutura_html(var_url):
    try:
        var_resposta = requests.get(var_url, timeout=15)
        var_resposta.raise_for_status()
        var_soup = BeautifulSoup(var_resposta.text, "html.parser")
        var_estrutura = var_extrair_estrutura_tag(var_soup.html)
        return var_estrutura
    except Exception as var_e:
        print(f"Erro ao processar {var_url}: {var_e}")
        return None

# Obter estruturas das URLs
var_estruturas = {}
for var_ano, var_url in var_urls.items():
    print(f"Processando ano {var_ano}...")
    var_estrutura = var_obter_estrutura_html(var_url)
    if var_estrutura:
        var_estruturas[var_ano] = var_estrutura

var_chaves = sorted(var_estruturas.keys())
var_n = len(var_chaves)

if var_n < 2:
    print("Poucos documentos para comparação.")
else:
    var_distancias = np.zeros((var_n, var_n))
    for i in range(var_n):
        for j in range(i + 1, var_n):
            var_ei = var_estruturas[var_chaves[i]]
            var_ej = var_estruturas[var_chaves[j]]
            var_s = difflib.SequenceMatcher(None, var_ei, var_ej).ratio()
            var_distancias[i, j] = 1 - var_s
            var_distancias[j, i] = var_distancias[i, j]

    var_dist_condensada = squareform(var_distancias)
    var_linkagem = linkage(var_dist_condensada, method='average')
```

```python
""" Dendograma - Diagrama ou representação icónica que organiza determinados fatores e variáveis num arranjo de agrupamentos derivado da 
aplicação de um "algoritmo de clustering"
"""

import numpy as np
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import pdist
from collections import defaultdict
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
from rich.console import Console
from rich.table import Table

def dendrograma_e_clusters(var_linkagem, var_chaves, threshold=0.5, caminho_save=None, salvar_figura=False, nome_figura="dendrograma.tif"):
    # Cria figura e dendrograma
    fig, ax = plt.subplots(figsize=(12, 6))

    dendrogram(
        var_linkagem,
        labels=var_chaves,
        leaf_rotation=90,
        leaf_font_size=10,
        color_threshold=threshold,
        orientation='top',
        truncate_mode='lastp',
        p=20,
        show_contracted=True,
        ax=ax
    )

    ax.set_title("Dendrograma das Estruturas HTML dos Congressos SBSR", fontsize=12)
    ax.text(
        0.5, 1.05,
        "Análise complementar dos clusters",
        fontsize=12,
        ha='center',
        va='bottom',
        transform=ax.transAxes
    )
    ax.set_xlabel("Ano dos Simpósios", fontsize=10)
    ax.set_ylabel("Distância (1 - Similaridade)")

    plt.tight_layout(pad=5)

    # Salvar figura se parâmetro ativado e caminho informado
    if salvar_figura and caminho_save:
        path = Path(caminho_save)
        path.mkdir(parents=True, exist_ok=True)
        arquivo_figura = path / nome_figura
        fig.savefig(arquivo_figura, format='tiff', dpi=300)
        print(f"Dendrograma salvo em: {arquivo_figura}")

    plt.show()

    # Formar clusters com limite definido
    clusters = fcluster(var_linkagem, threshold, criterion='distance')

    # Agrupar anos por cluster
    cluster_dict = defaultdict(list)
    for ano, cluster_id in zip(var_chaves, clusters):
        cluster_dict[cluster_id].append(ano)

    # DataFrame do resultado
    df_cluster = pd.DataFrame([
        {
            "id": cluster_id,
            "anos_clusterizados": ", ".join(sorted(cluster_anos)),
            "total": len(cluster_anos)
        }
        for cluster_id, cluster_anos in sorted(cluster_dict.items())
    ])

    # Salvar CSV e TSV se caminho existe
    if caminho_save:
        df_cluster.to_csv(path / "clusters_simposios.csv", index=False, encoding='utf-8-sig')
        df_cluster.to_csv(path / "clusters_simposios.tsv", index=False, sep='\t', encoding='utf-8-sig')

    # Mostrar tabela rica formatada
    console = Console()
    table = Table(show_header=True, header_style="bold black on yellow")
    table.add_column("id", style="dim", width=6)
    table.add_column("anos_clusterizados", overflow="fold")
    table.add_column("total", justify="right")

    for _, row in df_cluster.iterrows():
        style = "on grey93" if row["id"] % 2 == 0 else "on black"
        table.add_row(str(row["id"]), row["anos_clusterizados"], str(row["total"]), style=style)

    console.print(table)

    return df_cluster

# --- Exemplo de uso ---

var_chaves = ['2001', '2003', '2005', '2007', '2009', '2011', '2013', '2015', '2017', '2019']

np.random.seed(42)
data_simulada = np.random.rand(len(var_chaves), 4)  # 4 características

dist_matrix = pdist(data_simulada, metric='euclidean')
var_linkagem = linkage(dist_matrix, method='ward')

var_caminho_Trabalho_04_jupyter = r"C:\Users\profs\ICD_Trabalho_04_Análise_dos_Trabalhos_do_SBSR"

df_resultado = dendrograma_e_clusters(
    var_linkagem,
    var_chaves,
    threshold=3,
    caminho_save=var_caminho_Trabalho_04_jupyter,
    salvar_figura=True,
    nome_figura="dendrograma.tif"
)
```

```python
""" Rodando salva meta_tags em: csv, html e json
"""
import requests
from bs4 import BeautifulSoup
import pandas as pd
from IPython.display import display
import featuretools as ft
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar pandas para mostrar texto completo
pd.set_option('display.max_colwidth', None)

# URL para capturar meta tags
url = "http://marte2.sid.inpe.br/col/sid.inpe.br/marte2/2023/05.18.12.26/doc/thisInformationItemHomePage.html"

try:
    # Extração das meta tags
    response = requests.get(url)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, "html.parser")
    meta_tags = soup.find_all("meta")

    # Criar lista de dicionários com atributos
    meta_list = []
    for meta in meta_tags:
        meta_list.append({
            'name': meta.get('name'),
            'content': meta.get('content'),
          #  'property': meta.get('property'),
          #  'http-equiv': meta.get('http-equiv')
        })

    # Criar DataFrame
    df_meta = pd.DataFrame(meta_list)
    print("DataFrame original com meta tags:")
    display(df_meta)

    # Exportar arquivos
    df_meta.to_csv("meta_tags.csv", sep=';', index=False, encoding='utf-8-sig')
    print("CSV salvo em 'meta_tags.csv'.")

    df_meta.to_json("meta_tags.json", orient='records', force_ascii=False, indent=4)
    print("JSON salvo em 'meta_tags.json'.")

    df_meta.to_xml("meta_tags.xml", root_name='meta_tags', row_name='meta_tag', index=False, encoding='utf-8')
    print("XML salvo em 'meta_tags.xml'.\n")

    # Ler e exibir arquivos para validar
    df_csv = pd.read_csv("meta_tags.csv", sep=';')
    print("DataFrame carregado do CSV:")
    display(df_csv)

    df_json = pd.read_json("meta_tags.json", orient='records')
    print("DataFrame carregado do JSON:")
    display(df_json)

    df_xml = pd.read_xml("meta_tags.xml")
    print("DataFrame carregado do XML:")
    display(df_xml)

    # Preparar DataFrame para FeatureTools
    df_meta_ft = df_meta.reset_index().rename(columns={'index': 'id'})

    es = ft.EntitySet(id="meta_tags_data")
    es = es.add_dataframe(dataframe_name="meta",
                          dataframe=df_meta_ft,
                          index='id')

    # Gerar features automaticamente
    feature_matrix, feature_defs = ft.dfs(
        entityset=es,
        target_dataframe_name="meta",
        max_depth=1
    )
    print("\nFeature matrix gerada (primeiras linhas):")
    display(feature_matrix.head())

    # Visualizar distribuição do comprimento do conteúdo
    length_feature = None
    for f in feature_defs:
        if "length" in f.get_name().lower() and "content" in f.get_name().lower():
            length_feature = f.get_name()
            break

    if length_feature and length_feature in feature_matrix.columns:
        plt.figure(figsize=(8, 5))
        sns.histplot(feature_matrix[length_feature].dropna(), bins=20, kde=True)
        plt.title(f'Distribuição do comprimento das strings na coluna "content" ({length_feature})')
        plt.xlabel('Comprimento')
        plt.ylabel('Frequência')
        plt.show()
    else:
        print("Feature de comprimento para 'content' não encontrada.")

    # Mapa de calor das correlações
    numeric_features = feature_matrix.select_dtypes(include=['number'])
    plt.figure(figsize=(10, 8))
    sns.heatmap(
        numeric_features.corr(),
        annot=True,
        cmap='coolwarm',
        fmt=".2f",
        square=True
    )
    plt.title('Mapa de Calor das Correlações entre Features Numéricas')
    plt.show()

except requests.RequestException as e:
    print(f"Erro na requisição: {e}")
except FileNotFoundError as e:
    print(f"Arquivo não encontrado: {e}")
except Exception as e:
    print(f"Ocorreu um erro: {e}")
```
