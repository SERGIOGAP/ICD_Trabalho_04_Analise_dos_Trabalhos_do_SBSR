```python
""" DESCRIÇÃO GRAFO 01 - GRAFOS LINKS EXTERNOS - url_inicial = "https://www.sbsr.com.br/sbsr-2027/page/8162-inicio"

Para visualizar links externos 

Características:  gerar uma figura formato tiff a cores, representando um link graph, através de um scraping 
de dados de todas as url associadas à url: http://marte.sid.inpe.br/rep/83LX3pFwXQZ5Jpy/CxGU3, onde cada url é uma bolinha
vermelha com uma sequência numerada indicando cada url, uma legenda constanto número e o link de cada url

renomear estas url_inicial para cada grafo
"""
```

```python
### Grafo 01 - var_url_inicial = "https://www.sbsr.com.br/sbsr-2027/page/8162-inicio"

# url usada nesta célula - var_url_grafo_01 = var_url_inicial

# Define o caminho fixo para salvar os arquivos
#var_caminho_grafo_01 = caminho_Trabalho_04_jupyter

# Cria a pasta caso não exista
var_caminho_grafo_01.mkdir(parents=True, exist_ok=True)

def obter_pasta_downloads():
    var_home = os.path.expanduser("~")
    var_downloads = os.path.join(var_home, "Downloads")
    if not os.path.exists(var_downloads):
        return var_home
    return var_downloads

def coletar_urls(var_url_grafo_01):
    var_urls = set()
    try:
        var_resposta = requests.get(var_url_grafo_01)
        var_resposta.raise_for_status()
        var_soup = BeautifulSoup(var_resposta.text, 'html.parser')
        var_urls.add(var_url_grafo_01)
        for var_link in var_soup.find_all('a', href=True):
            var_href = var_link['href']
            if var_href.startswith('http'):
                var_urls.add(var_href)
    except Exception as var_e:
        print(f"Erro ao acessar {var_url_grafo_01}: {var_e}")
    return list(var_urls)

def construir_grafo(var_urls):
    var_G = nx.DiGraph()
    for var_i, var_url in enumerate(var_urls):
        var_G.add_node(var_i, url=var_url)
    for var_i in range(1, len(var_urls)):
        var_G.add_edge(0, var_i)
    return var_G

def salvar_urls_em_arquivos(var_urls):
    global var_df_grafo_01_urls
    var_df_grafo_01_urls = pd.DataFrame({
        'id': range(len(var_urls)),
        'url': var_urls
    })
    # Usa caminho absoluto da pasta definida
    var_caminho_csv = var_caminho_grafo_01 / "df_grafo_01_urls.csv"
    var_caminho_tsv = var_caminho_grafo_01 / "df_grafo_01_urls.tsv"

    # Salva arquivos CSV e TSV
    var_df_grafo_01_urls.to_csv(var_caminho_csv, index=False, encoding='utf-8')
    var_df_grafo_01_urls.to_csv(var_caminho_tsv, sep='\t', index=False, encoding='utf-8')

    print(f"URLs salvas em CSV: {var_caminho_csv}")
    print(f"URLs salvas em TSV: {var_caminho_tsv}")

    # Configurar pandas para imprimir o DataFrame completo
    pd.set_option('display.max_rows', None)
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    pd.set_option('display.colheader_justify', 'center')

    print("\nDataFrame CSV/tabulado completo formatado:\n")
    print(var_df_grafo_01_urls.to_string(index=False))

def desenhar_grafo(var_G):
    var_pos = nx.spring_layout(var_G, seed=42)
    plt.figure(figsize=(14, 8))

    var_nodes = var_G.nodes()
    nx.draw_networkx_nodes(var_G, var_pos, node_color='red', node_size=800)
    var_labels = {var_node: str(var_node) for var_node in var_nodes}
    nx.draw_networkx_labels(var_G, var_pos, var_labels, font_color='white', font_weight='bold')
    nx.draw_networkx_edges(var_G, var_pos, arrows=True)

    var_urls = nx.get_node_attributes(var_G, 'url')
    var_legenda = "Legenda:\n\n"
    for var_i in sorted(var_urls.keys()):
        var_legenda += f"{var_i}: {var_urls[var_i]}\n"

    plt.gca().text(1.05, 0.5, var_legenda, fontsize=10, va='center', ha='left', transform=plt.gca().transAxes,
                   bbox=dict(facecolor='yellow', edgecolor='black', boxstyle='round,pad=1'))

    plt.title("Grafo 01 - url principal do SBSR e links externos")
    plt.axis('off')
    plt.tight_layout()

    # Salvar o arquivo contendo o grafo na pasta definida
    var_caminho_arquivo = var_caminho_grafo_01/"grafo_01_url_principal_SBSR_links_externos.tiff"
    plt.savefig(var_caminho_arquivo, format='tiff', dpi=300)
    print(f"Arquivo salvo com sucesso em: {var_caminho_arquivo}")
    plt.show()

# Executa o processo
var_urls = coletar_urls(var_url_grafo_01)
salvar_urls_em_arquivos(var_urls)
var_grafo = construir_grafo(var_urls)
desenhar_grafo(var_grafo)
```

```python
### Grafo 03 - var_url_SBSR_bib_dig = "http://marte.sid.inpe.br/rep/83LX3pFwXQZ5Jpy/CxGU3"

# url usada nesta célula - var_url_grafo_03 = var_url_SBSR_bib_dig

# Define o caminho fixo para salvar os arquivos
#var_caminho_grafo_03 = caminho_Trabalho_04_jupyter

# Cria a pasta caso não exista
var_caminho_grafo_03.mkdir(parents=True, exist_ok=True)

def obter_pasta_downloads():
    var_home = os.path.expanduser("~")
    var_downloads = os.path.join(var_home, "Downloads")
    if not os.path.exists(var_downloads):
        return var_home
    return var_downloads

def eh_link_interno(var_url_grafo_03, var_link):
    var_base_dominio = urlparse(var_url_grafo_03).netloc
    var_link_parsed = urlparse(var_link)
    return (var_link_parsed.netloc == "" or var_link_parsed.netloc == var_base_dominio)

def coletar_links_internos(var_url_grafo_03):
    var_urls = set()
    try:
        var_resposta = requests.get(var_url_grafo_03)
        var_resposta.raise_for_status()
        var_soup = BeautifulSoup(var_resposta.text, 'html.parser')
        var_urls.add(var_url_grafo_03)
        for var_tag in var_soup.find_all('a', href=True):
            var_href = var_tag['href']
            var_href_completo = urljoin(var_url_grafo_03, var_href)  # Normaliza URL relativa
            if eh_link_interno(var_url_grafo_03, var_href_completo):
                var_urls.add(var_href_completo)
    except Exception as var_e:
        print(f"Erro ao acessar {var_url_grafo_03}: {var_e}")
    return list(var_urls)

def construir_grafo(var_urls):
    var_G = nx.DiGraph()
    for var_i, var_url in enumerate(var_urls):
        var_G.add_node(var_i, url=var_url)
    # Edges do nó inicial para os demais (simplificação)
    for var_i in range(1, len(var_urls)):
        var_G.add_edge(0, var_i)
    return var_G

def salvar_urls_em_arquivos(var_urls):
    global var_df_grafo_03_urls
    var_df_grafo_03_urls = pd.DataFrame({
        'id': range(len(var_urls)),
        'url': var_urls
    })
    # Usa caminho absoluto da pasta definida
    var_caminho_csv = var_caminho_grafo_03 / "df_grafo_03_urls.csv"
    var_caminho_tsv = var_caminho_grafo_03 / "df_grafo_03_urls.tsv"

    # Salva arquivos CSV e TSV
    var_df_grafo_03_urls.to_csv(var_caminho_csv, index=False, encoding='utf-8')
    var_df_grafo_03_urls.to_csv(var_caminho_tsv, sep='\t', index=False, encoding='utf-8')

    print(f"URLs salvas em CSV: {var_caminho_csv}")
    print(f"URLs salvas em TSV: {var_caminho_tsv}")

    # Configurar pandas para imprimir o DataFrame completo
    pd.set_option('display.max_rows', None)
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    pd.set_option('display.colheader_justify', 'center')

    print("\nDataFrame CSV/tabulado completo formatado:\n")
    print(var_df_grafo_03_urls.to_string(index=False))

def desenhar_grafo(var_G):
    var_pos = nx.spring_layout(var_G, seed=42)
    plt.figure(figsize=(14, 8))

    var_nodes = var_G.nodes()
    nx.draw_networkx_nodes(var_G, var_pos, node_color='red', node_size=800)
    var_labels = {var_node: str(var_node) for var_node in var_nodes}
    nx.draw_networkx_labels(var_G, var_pos, var_labels, font_color='white', font_weight='bold')
    nx.draw_networkx_edges(var_G, var_pos, arrows=True)

    var_urls = nx.get_node_attributes(var_G, 'url')
    var_legenda = "Legenda:\n"
    for var_i in sorted(var_urls.keys()):
        var_legenda += f"{var_i}: {var_urls[var_i]}\n"

    plt.gca().text(1.05, 0.5, var_legenda, fontsize=10, va='center', ha='left', transform=plt.gca().transAxes,
                   bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=1'))

    plt.title("grafo 03 - url da biblioteca digital do SBSR  e Links Internos")
    plt.axis('off')
    plt.tight_layout()

    var_caminho_arquivo = var_caminho_grafo_03 / "grafo_03_url_Biblioteca_Digital_SBSR_links_internos.tiff"

    plt.savefig(var_caminho_arquivo, format='tiff', dpi=300)
    print(f"Arquivo salvo com sucesso em: {var_caminho_arquivo}")
    plt.show()

# Executa o processo

var_urls = coletar_links_internos(var_url_grafo_03)
salvar_urls_em_arquivos(var_urls)

var_grafo = construir_grafo(var_urls)
desenhar_grafo(var_grafo)
```
